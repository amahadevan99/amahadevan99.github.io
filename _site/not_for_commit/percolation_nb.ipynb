```python
def f(p):
    return p**4 + 4*p**3*(1-p) + 2*p**2*(1-p)**2
```

```python
for p0 in linspace(0,1,30):
    x = [p0]
    for i in range(10):
        x.append(f(x[-1]))
    plot(x,linestyle='-',color='C0',linewidth=1)
xlabel('iterations ($n$)')
ylabel('effective $p$')
axhline(.5*(sqrt(5)-1),linestyle='--',color='gray',linewidth=1)
savefig('percolation_flows.png',bbox_inches='tight')
```
