```python
%pylab inline
```

```python
def f(K):
    return K**4 + 2*K**2 + 2*K**3
```

```python
for p0 in linspace(0,.5,15):
    x = [p0]
    for i in range(6):
        x.append(f(x[-1]))
    plot(x,linestyle='-',color='C0',linewidth=1)
xlabel('iterations')
ylabel('effective $K$')
axhline(.35321,linestyle='--',color='gray',linewidth=1)
ylim(0,1)
savefig('saw_rg_flows.png',bbox_inches='tight')
show()
```
